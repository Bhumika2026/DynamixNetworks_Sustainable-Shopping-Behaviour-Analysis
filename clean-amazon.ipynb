```python
import re

repeated_word_titles = df[
    df["Title"].str.contains(
        r'\b(\w+)\s+\1\b',
        case=False,
        regex=True,
        na=False
    )
]

repeated_word_titles[["Title"]].head(20)
```

```python
import re

def remove_repeated_words(text):
    if pd.isna(text):
        return text
    return re.sub(r'\b(\w+)(\s+\1\b)+', r'\1', text, flags=re.IGNORECASE)

df["Title"] = df["Title"].apply(remove_repeated_words)
```
